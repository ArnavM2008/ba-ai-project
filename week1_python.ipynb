```python
import pandas as pd

url = "https://raw.githubusercontent.com/ArnavM2008/ba-ai-project/refs/heads/main/Brazilian_Cup_Matches.csv"
df = pd.read_csv(url)
df.head()
```

```python
df.info()
```

```python
print(df.duplicated().sum(), "duplicates")
print(df.isna().sum())

df = df.drop_duplicates()
df = df.dropna()
print(df.shape)
```

```python
df["goal_difference"] = df["home_goal"] - df["away_goal"]

df.groupby("season")["goal_difference"].mean().plot(
    kind = "bar",
    title = "Average Home-Away Goal Difference by Season",
    xlabel = "Season",
    ylabel = "Average Goal Difference"
)

```
