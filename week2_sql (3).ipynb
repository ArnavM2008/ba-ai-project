```python
import pandas as pd, sqlite3
df = pd.read_csv("https://raw.githubusercontent.com/ArnavM2008/ba-ai-project/refs/heads/main/Brazilian_Cup_Matches.csv")
df = df.drop_duplicates().dropna()
df.columns = (df.columns.str.strip()
.str.lower().str.replace(" ", "_"))
```

```python
conn = sqlite3.connect(":memory:")
df.to_sql("data", conn, index=False)
def q(sql): return pd.read_sql(sql, conn)
```

```python
q("""
SELECT COUNT(*) AS home_wins
FROM data
WHERE home_goal > away_goal
""")
```



The home team won 579 matches.




```python
q("""
SELECT season, AVG(home_goal - away_goal) AS avg_goal_difference
FROM data
GROUP BY season
ORDER BY avg_goal_difference DESC
""")
```

The 2014 season had the greatest average home away goal difference with .497 goals per match.


```python
q("""
SELECT home_team, SUM(home_goal - away_goal) AS total_goal_difference
FROM data
WHERE season = 2020
GROUP BY home_team
ORDER BY total_goal_difference DESC
LIMIT 5
""")
```

During season 2020, Palmeiras - SP had the highest total home to away goal difference of 6, followed by Fluminense - RJ, Ferroviária, and Ceará with differences of 6 goals each, and Brusque with 4.
