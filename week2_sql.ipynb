```python
import pandas as pd, sqlite3
df = pd.read_csv("https://raw.githubusercontent.com/ArnavM2008/ba-ai-project/refs/heads/main/Brazilian_Cup_Matches.csv")
df = df.drop_duplicates().dropna()
df.columns = (df.columns.str.strip()
.str.lower().str.replace(" ", "_"))
```

```python
conn = sqlite3.connect(":memory:")
df.to_sql("data", conn, index=False)
def q(sql): return pd.read_sql(sql, conn)
```

Q2 - Which season had the most ties in matches?

Q3 - What are the top 5 home teams that had the highest total goals in a specific season?

```python
df.columns
```

```python
q("""
SELECT COUNT(*) AS orders
FROM data
WHERE home_goal + away_goal >= 4
""")
```

278 matches had 4 or more total goals.

```python
q("""
SELECT season, COUNT(*) AS ties
FROM data
WHERE home_goal = away_goal
GROUP BY season
ORDER BY ties DESC
LIMIT 1
""")
```

The 2016 season had the most ties with 50 ties.

```python
q("""
SELECT home_team, SUM(home_goal) AS total_goals
FROM data
WHERE season = 2020
GROUP BY home_team
ORDER BY total_goals DESC
LIMIT 5
""")
```

During season 2020, the top 5 home teams scoring the most goals were Ceará, Ferroviária, Brusque, Palmeiras, and Atlético
